```python
import pandas
import geopandas
import shapely
import pandas as pd
import geopandas as gpd
from shapely import Point
```

```python
#membuat kamus
#contoh data dari 5 kota besar di berbagai benua
data = {
    "Name": ["New York City", "São Paulo", "Tokyo", "Lagos", "Sydney"],
    "Population": [8419600, 12325232, 13929286, 15000000, 5312163],
    "Latitude": [40.7128, -23.5505, 35.6895, 6.5244, -33.8688],
    "Longitude": [-74.0060, -46.6333, 139.6917, 3.3792, 151.2093],
}
```

```python
type(data) #cara mengetahui tipe data
```

```python
#membuat dataframe
cities_df = pd.DataFrame(data)
```

```python
cities_df["Name"] #ambil data khusus dari dataframe
```

```python
gdf = gpd.GeoDataFrame( #gdf = geo data frame
    cities_df,
    geometry=gpd.points_from_xy(
        cities_df['Longitude'],
        cities_df['Latitude']
    )
)
```
